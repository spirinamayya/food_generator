# Lunch picker implementation check

Timestamp: 2026-09-11 21:54 +04

## Hypothesis and method

The static page should map every possible `Math.random()` index to the expected lunch name and matching image, then expose the selected result after a button click. The inline JavaScript was executed with a small mocked DOM in macOS JavaScript for Automation. Each of the six random intervals was forced once. The six external image URLs were also requested to confirm that they are reachable.

```bash
# Executed from the project root with the system JavaScript runtime.
# The full inline script was loaded from index.html into a mocked document.
osascript -l JavaScript -e '<load index.html; mock querySelector and classList; force each of 6 Math.random intervals; click button; assert label, image ID, and hidden placeholder>'
```

```bash
ruby -EUTF-8:UTF-8 -e 'html=File.read("index.html", encoding: "UTF-8"); urls=html.scan(%r{https://images\.unsplash\.com/[^\"]+}).uniq; abort "No photo URLs found" if urls.empty?; urls.each { |url| abort "Photo check failed: #{url}" unless system("curl", "-L", "--fail", "--silent", "--show-error", "--output", "/dev/null", "--max-time", "30", url) }; puts "PASS: all #{urls.length} lunch photo URLs responded successfully."'
```

## Result interpretation

The random-index logic, exact result message, placeholder transition, and name-to-photo mapping passed for every configured lunch. All configured photos were reachable at verification time. The page is a root-level, dependency-free `index.html`, so it is compatible with GitHub Pages static hosting.
